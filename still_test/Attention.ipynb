```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python


# 1. Generate synthetic data: 500 subjects, 10 random features
num_subjects = 500
num_features = 5
X_raw = torch.randn(num_subjects, num_features)  # Shape: (500, 10)

# 2. Create a column of ones for the first feature
ones_column = torch.ones(num_subjects, 1)  # Shape: (500, 1)

# 3. Concatenate the ones column to the front of the matrix 
# This changes the final shape of X to (500, 11)
X = torch.cat((ones_column, X_raw), dim=1)

# 4. Transpose data so the 11 features act as the "sequence"
X_features = X.t()  # Shape: (11, 500)

# 5. Define the projection dimension (d_k) 
d_k = 32
W_Q = nn.Linear(num_subjects, d_k, bias=False)
W_K = nn.Linear(num_subjects, d_k, bias=False)

# 6. Project features into Query (Q) and Key (K) spaces
Q = W_Q(X_features)  # Shape: (11, d_k)
K = W_K(X_features)  # Shape: (11, d_k)

# 7. Compute the raw attention scores
# (11, d_k) multiplied by (d_k, 11) results in a (11, 11) shape
scores = torch.matmul(Q, K.transpose(-2, -1))

# 8. Scale by sqrt(d_k) and apply Softmax row-wise
attention_matrix = F.softmax(scores / (d_k ** 0.5), dim=-1)

print("Attention Matrix Shape:", attention_matrix.shape)
print("\nAttention Matrix:\n", attention_matrix)
```
